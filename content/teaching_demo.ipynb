## Assigning Variables

Look at this code:

```python
customer = "Paul"

product_bought = "Laptop"

amount_spent = 10_000
```

Each line of this code is a *variable assignment statement*. 

When we use an assignment statement, we make a new variable. This saves information the memory of our computer.

Look at the first line of the code in the cell above:

![](https://raw.githubusercontent.com/pxr687/teaching_demo/refs/heads/main/content/data/look_at_this.png)

This *variable assignment statement* has two *sides*:

- The *left* of the `=`

- The *right* of the `=`

![](https://raw.githubusercontent.com/pxr687/teaching_demo/refs/heads/main/content/data/variable_assignment_left_right.png)

In this line of code, on the left of the `=` it says `customer`. This is the *name* of the variable.

On the right of the `=` it says `"Paul"`. This is the *value* of the variable.

![](https://raw.githubusercontent.com/pxr687/teaching_demo/refs/heads/main/content/data/variable_assignment_name_val.png)

The different colours tell us that we are looking at two different things: a *name* and a *value*.

Creating a variable means our computer stores the value in memory. We can use the name to get the value:

```python
customer
```

We can do this for all of the variables we created:

```python
product_bought
```

```python
amount_spent
```

## Using the `print()` function to show variable values

Now we have our data saved as variables, we can use the `print()` function to show all of our variables together in one message. We saw this function in the first code cell we used on this page.

In Python a *function* is code that is already written, which we use to do some task. The task of the `print()` function is to show us something. You can read `print()` as saying "show me this thing!":

![](https://raw.githubusercontent.com/pxr687/teaching_demo/refs/heads/main/content/data/print_function.png)

We can look at the value of just one variable like this:

```python
print(amount_spent)
```

**Tip:** look at the *colours* in the code below - some things are strings (red), some things are variables (black).

```python
# Print a message with the customer data.
print("Customer name:", customer, "| Product bought:", product_bought, "| Amount Spent:", amount_spent)
```

Remember that Paul came back to the shop and bought some speakers for 2000 baht?

Now we can easily *re-assign* the variables `product_bought` and `amount_spent` to store this new information. 

*Re-assigning a variable* means that we keep it's name, but we give it a new value. Here is the value of our variable BEFORE we re-assign it:

```python
# The value of our variable, BEFORE we re-assign it.
product_bought
```

In the cell below, we re-assign it. It has the same name, but a different value:

```python
# Re-assigning the variable same name, different value.
product_bought = product_bought + " and Speakers"

product_bought
```

We can add the cost of the speakers to the `amount_spent` variable, using variable re-assignment:

```python
# The value of our variable, BEFORE we re-assign it.
amount_spent
```

```python
# Re-assigning the variable: same name, different value.
amount_spent = amount_spent + 2000

amount_spent
```

We can now use the same variable names to show the updated information:

```python
print("Customer name:", customer, "| Product bought:", product_bought, "| Amount Spent:", amount_spent)
```

## Data Types & the `type()` function

Remember earlier I said to pay attention to the *colours* of things inside code cells?

This is because the colours tell us about the *type of data* we are using.

We have used text data and number data.

In Python, the name for text data is *string data*. We can use the `type()` function to tell us what data type we are looking at.

Remember that a *function* is code that is already written, which we use to do some task. The task of the `type()` function is to tell us *"what type of thing is this?"*.

We put what we want to ask about inside of brackets `()`, after the word `type`:

```python
type("Paul")
```

In the output of the cell above, `str` stands for "string". This means that `"Paul"` is text data.

![](https://raw.githubusercontent.com/pxr687/teaching_demo/refs/heads/main/content/data/type_function.png)

We can use `type()` to ask about anything in our Python code:

```python
type(10_000)
```

Here, `int` means "integer" which is a type of number data - integers are the *whole* numbers: 0, 1, 2, 3, 4 etc.

We can also ask what type of data the value of a variable is. Let's do this with our `customer` data:

```python
customer
```

```python
type(customer)
```

There are lots of other data types as well - you can read more about them [here](https://pxr687.github.io/Intro_Comp_Programming/1_08_00_Data_Types.html).

**Important:** you must know what type of data you are using! Not everything that works with one type of data will work with other types of data. If you do not pay attention to data type, you will get lots of errors in your code.

## Arithmetic Operators & Expressions

Above we saw *variable assignment*. We make a variable assignment using the `=` *operator*.

In Python, an operator is a symbol which does something special:

![](https://raw.githubusercontent.com/pxr687/teaching_demo/refs/heads/main/content/data/variable_assignment_all.png)

The *arithmetic operators* are symbols which we use to do maths and arithmetic.

We can use the `+` symbol to add literals together:

```python
10 + 10
```

This works with text (string) data as well as numbers:

```python
"Peter" + " Rush"
```

When we add number or text literals together like this, we make an *expression*.

An *expression* in Python code is *anything which returns a value*. If we run some code and we see some output below the code cell, then that code cell contains an expression:

```python
"Here are two strings" + " added together! This is an expression."
```

```python
# Here are two numbers added together, this is also an expression.
1999 + 2001
```

| Expression Type | Operator | Example    | Value     |
|-----------------|----------|------------|-----------|
| Addition        | `+`      | `2 + 3`    | `5`       |
| Subtraction     | `-`      | `2 - 3`    | `-1`      |
| Multiplication  | `*`      | `2 * 3`    | `6`       |
| Division        | `/`      | `100 / 10` | `10`      |
| Remainder       | `%`      | `101 % 10` | `1`       |
| Exponentiation (Power)  | `**`     | `10 ** 2`  | `

```python
# Try some out in this cell!
```

All of these operators work with number data. However, they do **NOT** all work with text (string) data:

```python
# This works.
"Computer" + " Science"
```

```python
# This also works.
"Ten times! " * 10
```

```python
# This does NOT work.
"Ouch" ** 2
```

Look carefully at what the error message says: `TypeError` - this means we have tried to do something that is not allowed for this data type. **Always read error messages carefully - this helps you to fix your code...**
